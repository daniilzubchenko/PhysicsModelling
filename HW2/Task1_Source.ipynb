```python
from dataclasses import dataclass
import numpy as np
import matplotlib.pyplot as plt
```

```python
def isotropic_directions(N=100):
    cos_theta = np.random.default_rng().uniform(-1, 1, N)
    sin_theta = np.sqrt(1-cos_theta**2)
    phi = np.random.default_rng().uniform(0, 2*np.pi, N)
    return np.array([np.cos(phi)*sin_theta, np.sin(phi)*sin_theta, cos_theta]).T
```

#### Источник с Пуассоновским числом фотонов
#### Источник, имеющий generator, выкидывающий направления фотонов банчами по photons_in_bunch штук

```python
@dataclass
class Source:
    """Source as generator of directions"""
    Y: int = 10000 # photons / MeV
    position: np.ndarray = np.array((0.,0.,0.))
    energy: float = 1.
    photons_in_bunch: int = 100
    
    def __post_init__(self):
        if isinstance(self.position, tuple): self.position = np.asarray(self.position)
        self.generator = self.make_photons()

    def make_photons(self):
        if self.energy > 0:
            N_photons = np.random.default_rng().poisson(self.Y * self.energy) # modified N_photons with Poisson distribution
            directions = isotropic_directions(N_photons)
            photons_obtained = 0
            while photons_obtained < N_photons:
                yield directions[photons_obtained:photons_obtained+self.photons_in_bunch]
                photons_obtained += self.photons_in_bunch
        else:
            raise ValueError(f"Input energy must be positive")
```
